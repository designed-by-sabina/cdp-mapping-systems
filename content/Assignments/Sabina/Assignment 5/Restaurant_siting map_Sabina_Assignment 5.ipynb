# Restaurant Sitting Mapping

Link to the website: https://github.com/designed-by-sabina/mapping_website


I am mapping types of restaurant sitting and proximity to it using various colors and bubble sizes. All NYC restaurants are mapped as black until you click on the map. Once that happens, the ones within one mile radius change color to show the sitting type and size to illustrate proximity to a chosen point.
